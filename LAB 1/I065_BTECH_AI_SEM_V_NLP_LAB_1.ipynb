# Part 1 : Text Preprocessing level one operations

```python
import nltk
import string

nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')

from nltk.tokenize import sent_tokenize, word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer
```

```python
text = "Hello there! This is Ninaad's NLP Lab, and it's Assignment 1. NLTK makes text pre-processing very easy, doesn't it?"
print(text)
```

### 1. Change of Case

```python
lower_text = text.lower()
upper_text = text.upper()

print("Lowercase:", lower_text)
print("Uppercase:", upper_text)
```

### 2. Sentence Tokenization

```python
sentences = sent_tokenize(lower_text)
print("Sentences:")
for i, s in enumerate(sentences, 1):
    print(i, s)
```

### 3. Word Tokenization

```python
words = word_tokenize(lower_text)
print("Words:")
print(words)
```

### 4. Punctuation Mark Removal

```python
words_no_punct = [w.strip(string.punctuation) for w in words]
words_no_punct = [w for w in words_no_punct if w]

print("Words without punctuation:")
print(words_no_punct)
```

### 5. Stop Word Removal

```python
stop_words = set(stopwords.words('english'))
words_no_stopwords = [w for w in words_no_punct if w not in stop_words]

print("Words without stopwords:")
print(words_no_stopwords)
```

### 6. Stemming

```python
stemmer = PorterStemmer()
stemmed_words = [stemmer.stem(w) for w in words_no_stopwords]

print("Stemmed words:")
print(stemmed_words)
```

### 7. Lemmatization

```python
lemmatizer = WordNetLemmatizer()
lemmatized_words = [lemmatizer.lemmatize(w) for w in words_no_stopwords]

print("Lemmatized words:")
print(lemmatized_words)
```

# Part 2 : Tokenization without using built-in NLTK functions

### 1. Custom Sentence Tokenization

```python
import re

def custom_sent_tokenize(input_text):
    input_text = input_text.strip()
    # split after a sentence-ending mark (. ! ?) followed by whitespace
    sentences = re.split(r'(?<=[.!?])\s+', input_text)
    return [s for s in sentences if s]

custom_sentences = custom_sent_tokenize(text)
print("Sentences:")
for i, s in enumerate(custom_sentences, 1):
    print(i, s)
```

### 2. Custom Word Tokenization

```python
def custom_word_tokenize(input_text):
    # match words (letters/digits/underscore, keeping contractions like "it's" together)
    # or a single punctuation character, as separate tokens
    return re.findall(r"\w+(?:'\w+)?|[^\w\s]", input_text)

custom_words = custom_word_tokenize(lower_text)
print("Words:")
print(custom_words)
```

# Part 3 : Stemming vs Lemmatization

| Point | Stemming | Lemmatization |
|---|---|---|
| What it does | Chops off the end of a word using simple rules, to get a root form | Uses a dictionary and grammar rules to find the actual base word (called the "lemma") |
| Output | May not be a real word (e.g. "studies" → "studi") | Is always a real, valid word (e.g. "studies" → "study") |
| Speed | Fast, since it's just rule-based cutting | Slower, since it needs to look words up and check meaning/part of speech |
| Accuracy | Less accurate, can be a bit crude | More accurate, gives correct base word |
| Needs context (POS)? | No | Often yes (knowing if a word is a noun/verb helps pick the right lemma) |
| Example | "caring" → "car" (wrong meaning) | "caring" → "care" (correct meaning) |
| Common tool in NLTK | `PorterStemmer` | `WordNetLemmatizer` |
| When to use | When speed matters more than perfect accuracy (e.g. search engines) | When correctness/readability matters (e.g. chatbots, question answering) |

# Part 6 : Rule-Based Text Classification (BBC News Dataset)

### 1. Load Dataset

Dataset: `df_file.csv` has two columns, `Text` and `Label` (label is a number 0-4).
Checking a few sample documents per label shows the category each number stands for:

| Label | Category |
|---|---|
| 0 | politics |
| 1 | sport |
| 2 | tech |
| 3 | entertainment |
| 4 | business |

```python
import pandas as pd

df = pd.read_csv("df_file.csv")

label_to_category = {
    0: "politics",
    1: "sport",
    2: "tech",
    3: "entertainment",
    4: "business",
}
df["Category"] = df["Label"].map(label_to_category)

print(df.shape)
print(df["Category"].value_counts())
df.head()
```

### 2. Define Rule-Based Keyword Sets

For each category, we list a set of keywords that commonly show up in that kind of news article.

```python
category_keywords = {
    "politics": [
        "government", "minister", "parliament", "election", "labour", "tory",
        "tories", "party", "mp", "vote", "policy", "blair", "brown", "cabinet",
        "prime minister", "chancellor", "political", "campaign", "referendum",
    ],
    "sport": [
        "match", "team", "player", "game", "coach", "league", "championship",
        "cup", "tournament", "goal", "score", "football", "cricket", "rugby",
        "olympic", "tennis", "athlete", "club", "manager", "injury",
    ],
    "tech": [
        "computer", "software", "internet", "technology", "digital", "website",
        "online", "device", "mobile", "broadband", "microsoft", "google",
        "gadget", "network", "user", "app", "data", "phone", "pc",
    ],
    "entertainment": [
        "film", "movie", "music", "actor", "actress", "award", "star",
        "album", "band", "director", "concert", "show", "celebrity", "singer",
        "hollywood", "tv", "drama", "festival",
    ],
    "business": [
        "company", "market", "economy", "shares", "firm", "profit", "business",
        "bank", "trade", "economic", "growth", "investment", "stock", "sales",
        "chief executive", "revenue", "financial", "prices",
    ],
}
```

### 3. Rule-Based Classifier

The rule: lowercase the document, count how many times each category's keywords appear in it, and pick the category with the highest count. If no keyword matches at all, fall back to the overall most common category.

```python
fallback_category = df["Category"].value_counts().idxmax()

def rule_based_classify(document):
    document = document.lower()
    scores = {
        category: sum(document.count(keyword) for keyword in keywords)
        for category, keywords in category_keywords.items()
    }
    best_category = max(scores, key=scores.get)
    if scores[best_category] == 0:
        return fallback_category
    return best_category

# quick check on one document
print(df["Text"].iloc[0][:100])
print("Predicted:", rule_based_classify(df["Text"].iloc[0]))
print("Actual:", df["Category"].iloc[0])
```

### 4. Apply Classifier to the Whole Dataset and Check Accuracy

```python
df["Predicted"] = df["Text"].apply(rule_based_classify)

accuracy = (df["Predicted"] == df["Category"]).mean()
print(f"Accuracy: {accuracy:.2%}")

print("\nConfusion matrix (rows = actual, columns = predicted):")
confusion = pd.crosstab(df["Category"], df["Predicted"])
print(confusion)
```
