# Part 2 Regex

# a. Sequence starting with "he", two of any character, then "o"

```python
txt = "hello world"
result = re.search(r"he..o", txt)   # . means "any single character"
print(result)   # matches "hello"
```

# b. Starts with / ends with checks

```python
txt = "The rain in Spain"

# starts with "The"
print(bool(re.match(r"^The", txt)))   # ^ means "start of string"

# ends with "Spain"
print(bool(re.search(r"Spain$", txt)))   # $ means "end of string"

# starts with "The" AND ends with "Spain"
print(bool(re.match(r"^The.*Spain$", txt)))   # .* means "anything in between"
```

# c. Find all lowercase/uppercase letters between a and m

```python
lower = re.findall(r"[a-m]", txt)   # lowercase a to m
upper = re.findall(r"[A-M]", txt)   # uppercase A to M

print(lower)
print(upper)
```

# Part 3 - Application

## Automated Resume Screening System


Given a job description file and a set of resumes, the system:

1. reads the job description **line by line** and parses out the required skills and the eligibility cutoffs
2. reads each resume and extracts the candidate's name, email, phone number, profile URL, CGPA and graduation year using regex
3. normalises the resume text with `re.sub` (strips URLs, emails, digits and punctuation, collapses whitespace, lowercases)
4. tokenises the normalised text with `re.findall(r"[a-z]+", text)`
5. scores each candidate on how many required skills they match
6. applies the eligibility rules and writes a ranked report to disk, then reads it back

**note:** Normalisation and tokenisation are exactly what NLTK's `word_tokenize` or spaCy's tokeniser do as well. Could be done in further labs using inbuilt library functions.

### Step 0 - Create the input data

Set of resumes

```python
job_description = """Job Role: NLP Engineer Intern
Company: Arrow Analytics
Location: Mumbai
Required Skills: python, regex, pandas, machine learning, sql, natural language processing
Minimum CGPA: 7.0
Minimum Graduation Year: 2026
"""

resumes = {
"resume_1.txt": """Ninaad Wagle
Email: ninaad.wagle@nmims.edu.in
Phone: +91 9876543210
GitHub: https://github.com/ArrowAnalytics
CGPA: 8.7
Graduation: 2027

Skills: Python, Pandas, NumPy, Regex, SQL, Machine Learning, Natural Language Processing
Projects: Built a volatility research pipeline in Python using Pandas and regex based log parsing.
""",

"resume_2.txt": """Aarav Mehta
Email: aarav.mehta@gmail.com
Phone: 9812345678
LinkedIn: https://linkedin.com/in/aaravmehta
CGPA: 6.4
Graduation: 2026

Skills: Python, SQL, Excel
Projects: Dashboard reporting using Excel and SQL queries.
""",

"resume_3.txt": """Sara Fernandes
Email: sara.fernandes@@nmims..com
Phone: 9123456780
GitHub: github.com/sarafernandes
CGPA: 9.1
Graduation: 2027

Skills: Python, Machine Learning, Natural Language Processing, Pandas, Regex
Projects: Sentiment classifier trained on product reviews.
""",

"resume_4.txt": """Rohan Iyer
Email: rohan.iyer@outlook.com
CGPA: 7.8
Graduation: 2025

Skills: Java, Spring Boot, SQL
Projects: Backend REST services for an inventory system.
""",

"resume_5.txt": """Ishita Rao
Email: ishita.rao@nmims.edu.in
Phone: +91-9988776655
LinkedIn: www.linkedin.com/in/ishitarao
CGPA: 8.2
Graduation: 2026

Skills: Python, Regex, SQL, Pandas, Natural Language Processing
Projects: Rule based chatbot built entirely with regular expressions.
"""
}

with open("job_description.txt", "w") as file:
    file.write(job_description)

for name, text in resumes.items():
    with open(name, "w") as file:
        file.write(text)

resume_files = list(resumes.keys())
print("Created job_description.txt and", len(resume_files), "resume files")
```

### Step 1 - Parse the job description

Read **line by line** and use `re.match` with the `^` anchor to identify each field, then `re.sub` to strip the label away and leave only the value.

```python
def parse_job_description(path):
    requirements = {}
    with open(path, "r") as file:
        for line in file:
            line = line.strip()
            if re.match(r"^Job Role:", line, re.IGNORECASE):
                requirements["role"] = re.sub(r"^Job Role:\s*", "", line, flags=re.IGNORECASE)
            elif re.match(r"^Required Skills:", line, re.IGNORECASE):
                value = re.sub(r"^Required Skills:\s*", "", line, flags=re.IGNORECASE)
                requirements["skills"] = [s.strip().lower() for s in value.split(",")]
            elif re.match(r"^Minimum CGPA:", line, re.IGNORECASE):
                requirements["min_cgpa"] = float(re.search(r"\d+\.?\d*", line).group())
            elif re.match(r"^Minimum Graduation Year:", line, re.IGNORECASE):
                requirements["min_year"] = int(re.search(r"20\d{2}", line).group())
    return requirements


requirements = parse_job_description("job_description.txt")

for key, value in requirements.items():
    print(key, ":", value)
```

### Step 2 - Extract structured details from a resume

Each field gets its own pattern:

| Field | Pattern | Idea |
|---|---|---|
| Email | `[\w.+-]+@\S+` | grab loosely, validate strictly afterwards |
| Phone | `(?:\+91[-\s]?)?\b\d{10}\b` | optional country code, then exactly 10 digits |
| Profile | `(?:https?://)?(?:www\.)?(?:linkedin\.com\|github\.com)/[\w\-./]+` | alternation with `\|`, optional scheme and `www.` |
| CGPA | `CGPA\s*:?\s*(\d+\.?\d*)` | capture group returns only the number |
| Year | `Graduation\s*:?\s*(20\d{2})` | anchored to the label so it cannot grab a project year |

The email validator here is a **corrected version** of the one from Part 2(f). That pattern was `^[\w.+-]+@[\w-]+\.[a-zA-Z]{2,}$`, which allows only a single dot in the domain, so a real address like `ninaad.wagle@nmims.edu.in` is wrongly rejected. Adding `(?:\.[\w-]+)*` lets the domain have any number of sub-levels.

```python
def is_valid_email(email):
    pattern = r"^[\w.+-]+@[\w-]+(?:\.[\w-]+)*\.[a-zA-Z]{2,}$"
    return bool(re.match(pattern, email))


def extract_details(text):
    details = {}
    lines = text.strip().split("\n")
    details["name"] = lines[0].strip()

    email = re.search(r"[\w.+-]+@\S+", text)
    details["email"] = email.group() if email else "Not found"
    details["email_valid"] = is_valid_email(details["email"])

    phone = re.search(r"(?:\+91[-\s]?)?\b\d{10}\b", text)
    details["phone"] = phone.group() if phone else "Not found"

    profile = re.search(r"(?:https?://)?(?:www\.)?(?:linkedin\.com|github\.com)/[\w\-./]+", text)
    details["profile"] = profile.group() if profile else "Not found"

    cgpa = re.search(r"CGPA\s*:?\s*(\d+\.?\d*)", text, re.IGNORECASE)
    details["cgpa"] = float(cgpa.group(1)) if cgpa else 0.0

    year = re.search(r"Graduation\s*:?\s*(20\d{2})", text, re.IGNORECASE)
    details["year"] = int(year.group(1)) if year else 0

    return details


with open("resume_1.txt", "r") as file:
    sample = file.read()

for key, value in extract_details(sample).items():
    print(key, ":", value)
```

### Step 3 - Normalise and tokenise

`clean_text` applies four substitutions in order, and the order matters: URLs and emails must go before punctuation is stripped, because once the dots and slashes are gone there is nothing left to recognise them by.

`tokenize` then splits the cleaned string into word tokens. This is a regex tokeniser, the same approach NLTK's `RegexpTokenizer` uses.

```python
def clean_text(text):
    text = re.sub(r"(?:https?://)?(?:www\.)?\S+\.(?:com|in|org|net)\S*", " ", text)
    text = re.sub(r"[\w.+-]+@[\w.-]+", " ", text)
    text = re.sub(r"[^A-Za-z\s]", " ", text)
    text = re.sub(r"\s+", " ", text)
    return text.strip().lower()


def tokenize(text):
    return re.findall(r"[a-z]+", text)


cleaned = clean_text(sample)
print("CLEANED TEXT")
print(cleaned)
print()
print("TOKENS")
print(tokenize(cleaned))
print()
print("TOKEN COUNT:", len(tokenize(cleaned)), " UNIQUE:", len(set(tokenize(cleaned))))
```

### Step 4 - Score a resume against the required skills

Single-word skills are checked against the **token set**, which is an exact match and avoids false positives (looking for `sql` in the raw string would also match inside `mysql`).

Multi-word skills such as `machine learning` cannot be found in a set of single tokens, so those are checked as a substring of the cleaned text instead. This is the point where a real system would move to n-grams.

```python
def score_resume(text, skills):
    cleaned = clean_text(text)
    tokens = set(tokenize(cleaned))
    matched = []
    for skill in skills:
        if " " in skill:
            if skill in cleaned:
                matched.append(skill)
        elif skill in tokens:
            matched.append(skill)
    missing = [s for s in skills if s not in matched]
    percent = round(len(matched) / len(skills) * 100, 2)
    return percent, matched, missing


percent, matched, missing = score_resume(sample, requirements["skills"])
print("Score   :", percent, "percent")
print("Matched :", matched)
print("Missing :", missing)
```

### Step 5 - Screen every resume and rank them

A candidate is shortlisted only if all four rules pass: CGPA above cutoff, graduating in or after the cutoff year, a validly formatted email, and a skill match of at least 50 percent.

```python
def screen(files, requirements):
    results = []
    for path in files:
        with open(path, "r") as file:
            text = file.read()

        details = extract_details(text)
        percent, matched, missing = score_resume(text, requirements["skills"])

        reasons = []
        if details["cgpa"] < requirements["min_cgpa"]:
            reasons.append("CGPA below cutoff")
        if details["year"] < requirements["min_year"]:
            reasons.append("graduated before cutoff year")
        if not details["email_valid"]:
            reasons.append("invalid email format")
        if percent < 50:
            reasons.append("skill match below 50 percent")

        details["file"] = path
        details["score"] = percent
        details["matched"] = matched
        details["missing"] = missing
        details["reasons"] = reasons
        details["status"] = "SHORTLISTED" if not reasons else "REJECTED"
        results.append(details)

    results.sort(key=lambda r: r["score"], reverse=True)
    return results


results = screen(resume_files, requirements)

for r in results:
    print(r["status"].ljust(12), str(r["score"]).rjust(6), " ", r["name"])
```

### Step 6 - Write the report to a file

```python
report_lines = []
report_lines.append("RESUME SCREENING REPORT")
report_lines.append("Role: " + requirements["role"])
report_lines.append("Required Skills: " + ", ".join(requirements["skills"]))
report_lines.append("Cutoffs: CGPA >= " + str(requirements["min_cgpa"]) + " | Graduation >= " + str(requirements["min_year"]))
report_lines.append("=" * 62)

for rank, r in enumerate(results, start=1):
    report_lines.append("Rank " + str(rank) + ": " + r["name"] + "  [" + r["status"] + "]")
    report_lines.append("  File     : " + r["file"])
    report_lines.append("  Email    : " + r["email"] + ("" if r["email_valid"] else "   <-- INVALID FORMAT"))
    report_lines.append("  Phone    : " + r["phone"])
    report_lines.append("  Profile  : " + r["profile"])
    report_lines.append("  CGPA     : " + str(r["cgpa"]) + "    Graduation: " + str(r["year"]))
    report_lines.append("  Match    : " + str(r["score"]) + " percent")
    report_lines.append("  Matched  : " + (", ".join(r["matched"]) if r["matched"] else "none"))
    report_lines.append("  Missing  : " + (", ".join(r["missing"]) if r["missing"] else "none"))
    if r["reasons"]:
        report_lines.append("  Rejected : " + "; ".join(r["reasons"]))
    report_lines.append("-" * 62)

with open("screening_report.txt", "w") as file:
    for line in report_lines:
        file.write(line + "\n")

print("Report written to screening_report.txt")
```

### Step 7 - Read the report back

```python
with open("screening_report.txt", "r") as file:
    print(file.read())
```

### Step 8 - Read a specific line from the report

`readlines()` returns a list, so any single line can be indexed directly. Line 6 is the top ranked candidate.

```python
with open("screening_report.txt", "r") as file:
    lines = file.readlines()

print("Total lines in report:", len(lines))
print("Top candidate:", lines[5].rstrip())
```

### Step 9 - Write the shortlist as a clean pipe separated file

This is the file a recruiter would actually open, and its format can be loaded straight into a spreadsheet.

```python
shortlist = [r for r in results if r["status"] == "SHORTLISTED"]

with open("shortlist.txt", "w") as file:
    file.write("Name | Email | Phone | Match\n")
    for r in shortlist:
        file.write(r["name"] + " | " + r["email"] + " | " + r["phone"] + " | " + str(r["score"]) + "\n")

with open("shortlist.txt", "r") as file:
    print(file.read())

print("Shortlisted:", len(shortlist), "out of", len(results))
```
